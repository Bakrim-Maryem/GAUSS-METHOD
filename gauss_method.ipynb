```python

import numpy as np

def gauss_pivot_partiel(A, B):
    A = np.array(A, dtype=float)
    B = np.array(B, dtype=float)

    n = len(B)

    for k in range(n - 1):

        p = k + np.argmax(np.abs(A[k:, k]))

        A[[k, p]] = A[[p, k]]
        B[[k, p]] = B[[p, k]]

        if abs(A[k, k]) < 1e-12:
            raise ValueError("Matrice singuliere ou pivot nul")

        for i in range(k + 1, n):
            m = A[i, k] / A[k, k]
            A[i, k:] = A[i, k:] - m * A[k, k:]
            B[i] = B[i] - m * B[k]

    if abs(A[n - 1, n - 1]) < 1e-12:
        raise ValueError("Matrice singuliere ou pivot nul")

    X = np.zeros(n)

    for i in range(n - 1, -1, -1):
        X[i] = (
            B[i] - np.dot(A[i, i + 1:], X[i + 1:])
        ) / A[i, i]

    return X

A = [[2, 1, -1],
     [-3, -1, 2],
     [-2, 1, 2]]

B = [8, -11, -3]

X = gauss_pivot_partiel(A, B)
print("Solution par pivot partiel :", X)
```

```python
import numpy as np

def gauss_pivot_total(A, B):
    A = np.array(A, dtype=float)
    B = np.array(B, dtype=float)

    n = len(B)

    ordre = np.arange(n)

    for k in range(n - 1):

        sous_matrice = np.abs(A[k:, k:])
        i_max, j_max = np.unravel_index(
            np.argmax(sous_matrice),
            sous_matrice.shape
        )

        i_max += k
        j_max += k

        A[[k, i_max]] = A[[i_max, k]]
        B[[k, i_max]] = B[[i_max, k]]

        A[:, [k, j_max]] = A[:, [j_max, k]]
        ordre[[k, j_max]] = ordre[[j_max, k]]

        if abs(A[k, k]) < 1e-12:
          raise ValueError("Matrice singuliere ou pivot nul")

        for i in range(k + 1, n):
            m = A[i, k] / A[k, k]
            A[i, k:] -= m * A[k, k:]
            B[i] -= m * B[k]

    if abs(A[n - 1, n - 1]) < 1e-12:
        raise ValueError("Matrice singuliere ou pivot nul")

    X_perm = np.zeros(n)

    for i in range(n - 1, -1, -1):
        X_perm[i] = (
            B[i] - np.dot(A[i, i + 1:], X_perm[i + 1:])
        ) / A[i, i]

    X = np.zeros(n)
    X[ordre] = X_perm

    return X
A = [[2, 1, -1],
     [-3, -1, 2],
     [-2, 1, 2]]

B = [8, -11, -3]

X = gauss_pivot_total(A, B)
print("Solution par pivot total :", X)
```
